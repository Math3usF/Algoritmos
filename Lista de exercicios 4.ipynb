```python
# 1. Soma dos Dígitos de um Número Inteiro
def soma_digitos(n):    #função recursiva para somar os dígitos de um número inteiro
    if n == 0:          #se o número for 0, a soma dos dígitos é 0
        return 0    #caso base: se o número for 0, a soma dos dígitos é 0
    return (n % 10) + soma_digitos(n // 10)    #Recursividade

# teste da função
numero = 123
resultado = soma_digitos(numero)
print(f"A soma dos digitos de {numero} é {resultado}")
```

```python
# 2. Contar Dígitos
def contar_digitos(n):    #função recursiva para contar os dígitos de um número inteiro
    if n < 10:             #caso base: se o número for menor que 10, ele tem apenas um dígito
        return 1
    return 1 + contar_digitos(n//10)    #Recursividade

# teste da função
numero = 9876
resultado = contar_digitos(numero)
print(f"O número de dígitos de {numero} é {resultado}")
```

```python
#3. Inverter uma String
def inverter_string(s):    #função recursiva para inverter uma string
    if s == "":            #caso base: se a string for vazia, retorna a string vazia
        return s
    return s[-1] + inverter_string(s[:-1])    #Recursividade

# teste da função
texto = "recursividade"
resultado = inverter_string(texto) #chama a função para inverter a string "recursividade" e armazena o resultado na variável resultado
if texto == "":
    print("A string está vazia.")
else:
    print(f"A string invertida de {texto} é {resultado}")
```

```python
#4. Verificar palíndromo
def verificar_palindromo(s):    #função recursiva para verificar se uma string é um palíndromo
    if len(s) <= 1:                   #caso base: se a string tiver 0 ou 1 caractere, é um palíndromo
        return True
    if s[0] != s[-1]:                 #se o primeiro e o último caractere forem diferentes, não é um palíndromo
        return False
    return verificar_palindromo(s[1:-1])    #Recursividade

# teste da função
print(f"'arara' é palíndromo? {verificar_palindromo('arara')   }") 
```

```python
#5. Produto usando soma

def produto(a,b):
    if b == 0:             #caso base: se b for 0, o produto é 0
        return 0
    if b < 0:               #se b for negativo, inverte os sinais e chama a função novamente
        return -produto(a, -b)
    return a + produto(a, b-1)    #Recursividade

#teste de função 
print(f"O produto de 3 e 4 é: {produto(3, 4)}")
```

```python
#6. Maior elemento de uma lista
def maior_elemento(lista):
    if len(lista) == 1:     #caso base: se a lista tiver apenas um elemento, esse é o maior
        return lista[0]     #se a lista tiver mais de um elemento, compara o primeiro elemento com o maior do restante da lista
    max_restante = maior_elemento(lista[1:])    #Recursividade
    return lista[0] if lista[0] > max_restante else max_restante

# teste da função
print(f"Maior de [1, 5, 3, 9, 2]: {maior_elemento([1, 5, 3, 9, 2])}")
```

```python
#7. MDC
def mdc(a, b):      
    if b == 0:             #caso base: se b for 0, o MDC é a
        return a
    return mdc(b, a % b)    #Recursividade segue o algoritmo de Euclides para calcular o MDC

print(f"O MDC de 48 e 18 é: {mdc(48, 18)}")
```

```python
#8. Contar Ocorrências de um Elemento em uma Lista
def contar_ocorrencias(lista, elemento):
    if not lista:          #caso base: se a lista estiver vazia, retorna 0
        return 0
    count_restante = contar_ocorrencias(lista[1:], elemento)    #Recursividade
    return (1 if lista[0] == elemento else 0) + count_restante

# teste da função
print(f"Ocorrências de 3 em [1, 3, 3, 7, 3]: o número 3 aparece {contar_ocorrencias([1, 3, 3, 7, 3], 3)} vezes ")
```

```python
#9. Soma de elemnentos de uma lista
def soma_lista(lista):
    if not lista:          #caso base: se a lista estiver vazia, retorna 0
        return 0            #se a lista não estiver vazia, soma o primeiro elemento com a soma do restante da lista
    return lista[0] + soma_lista(lista[1:])    #Recursividade aqui a função soma o primeiro elemento da lista com a soma do restante da lista

#Teste: 
print(f"A soma dos elementos de [1, 2, 3, 4, 5] é: {soma_lista([1, 2, 3, 4, 5])}")
```

```python
#10. Verificar se o número é primo
def eh_primo(n, divisor=2):
    if n < 2:              #caso base: números menores que 2 não são primos
        return False
    if divisor >= n:       #caso base: se o divisor for maior ou igual a n, n é primo
        return True
    if n % divisor == 0:   #se n for divisível por divisor, não é primo
        return False
    return eh_primo(n, divisor + 1)    #Recursividade

#teste da função
numero = int(input("Digite um número para verificar se é primo: "))
if eh_primo(numero):
    print(f"{numero} é um número primo.")
else:    print(f"{numero} não é um número primo.")  
```
