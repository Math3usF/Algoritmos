```python
def indice_do_menor(lista):
    
    if not lista:
        return -1

    indice_menor = 0 
    
    for i in range(1, len(lista)):
        
        if lista[i] < lista[indice_menor]:
            indice_menor = i  
            
    return indice_menor
precos = [25.50, 10.90, 45.00, 8.25, 15.00]
posicao = indice_do_menor(precos)

print(f"O menor valor está no índice: {posicao}") 
```

```python
def busca_nome(lista, alvo):
    alvo_lower = alvo.lower()
    for i, nome in enumerate(lista):
        if nome.lower() == alvo_lower:
            return i
    return -1

lista_nomes = ["Enzo", "ALVEZ", "Maria", "moises"]
# Índice:       0        1          2         3


print(busca_nome(lista_nomes, "Alvez"))   
print(busca_nome(lista_nomes, "Enzo"))   
print(busca_nome(lista_nomes, "Moises")) 
print(busca_nome(lista_nomes, "Cole"))   
```

```python
def busca_matriz(matriz, alvo):
    # O primeiro loop
    for i in range(len(matriz)):
        # O segundo loop 
        for j in range(len(matriz[i])):
            if matriz[i][j] == alvo:
                return (i, j)  # (linha, coluna)
                
    return (-1, -1)  #retorna erro se terminar os dois loops e não encontrar
minha_matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

alvo = 9
posicao = busca_matriz(minha_matriz, alvo)

print(f"O número {alvo} está na posição: {posicao}") 
```

```python
import random

def busca_com_limite(lista):
    
    for i in range(len(lista)):
        if lista[i] > 50:
            return i, lista[i]  
            
    return -1, None  # Caso não encontre


random.seed(42)
numeros_aleatorios = [random.randint(1, 100) for _ in range(100)]


indice, valor = busca_com_limite(numeros_aleatorios)

if indice != -1:
    print(f"Primeiro número > 50 encontrado no índice {indice}: valor {valor}")
else:
    print("Nenhum número maior que 50 foi encontrado na lista.")
```

```python
import time

# Busca Sequencial 
def busca_sequencial(lista, alvo):
    for i in range(len(lista)):
        if lista[i] == alvo:
            return i
    return -1

# Busca Binária
def busca_binaria(lista, alvo):
    inicio = 0
    fim = len(lista) - 1
    
    while inicio <= fim:
        meio = (inicio + fim) // 2
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
    return -1

#Comparação
lista_ordenada = list(range(10000000)) # 10 milhões
alvo = 9999999 #pior caso para busca sequencial

# Busca Sequencial
start = time.time()
busca_sequencial(lista_ordenada, alvo)
end = time.time()
print(f"Busca Sequencial: {end - start:.5f} segundos")

# Busca Binária
start = time.time()
busca_binaria(lista_ordenada, alvo)
end = time.time()
print(f"Busca Binária: {end - start:.5f} segundos")
```

```python
def busca_binaria(lista_ordenada, alvo):
    inicio = 0
    fim = len(lista_ordenada) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2  
        
        # Encontrou o alvo
        if lista_ordenada[meio] == alvo:
            return meio
        
        #  O alvo é maior que o valor do meio 
        elif lista_ordenada[meio] < alvo:
            inicio = meio + 1
            
        #  O alvo é menor que o valor do meio
        else:
            fim = meio - 1
            
    return -1  


minha_lista = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
alvo = 338

resultado = busca_binaria(minha_lista, alvo)

if resultado != -1:
    print(f"Alvo encontrado no índice: {resultado}") 
else:
    print("Alvo não encontrado.")
```

```python
def busca_primeira_ocorrencia(lista, alvo):
    inicio = 0
    fim = len(lista) - 1
    resultado = -1 
    
    while inicio <= fim:
        meio = (inicio + fim) // 2
        
        if lista[meio] == alvo:
            resultado = meio    # Encontrou ...
            fim = meio - 1     
            
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
            
    return resultado


lista_com_duplicatas = [1, 2, 2, 2, 3, 4, 4, 5]
alvo = 2

indice = busca_primeira_ocorrencia(lista_com_duplicatas, alvo)
print(f"O menor índice do número {alvo} é: {indice}") 
```

```python
def busca_ultima_ocorrencia(lista, alvo):
    inicio = 0
    fim = len(lista) - 1
    resultado = -1
    
    while inicio <= fim:
        meio = (inicio + fim) // 2
        
        if lista[meio] == alvo:
            resultado = meio   
            inicio = meio + 1   
            
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
            
    return resultado


lista_com_duplicatas = [1, 2, 2, 2, 3, 4, 4, 5]
alvo = 2

indice = busca_ultima_ocorrencia(lista_com_duplicatas, alvo)
print(f"O maior índice do número {alvo} é: {indice}") 
```

```python
def busca_posicao_insercao(lista, alvo):
    inicio = 0
    fim = len(lista) - 1
    
    while inicio <= fim:
        meio = (inicio + fim) // 2
        
        if lista[meio] == alvo:
            return meio  # se já existe deve retorna o índice atual
        
        if lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
            
   
    return inicio


lista_exemplo = [10, 20, 30, 40, 50]

# O número já existe
print(busca_posicao_insercao(lista_exemplo, 30)) 

# O número não existe 
print(busca_posicao_insercao(lista_exemplo, 25)) 

# O número é maior que todos
print(busca_posicao_insercao(lista_exemplo, 60)) 
```

```python
def busca_binaria_nome(lista_nomes, alvo):
    inicio = 0
    fim = len(lista_nomes) - 1
    alvo_normalizado = alvo.lower() # Padroniza o alvo 

    while inicio <= fim:
        meio = (inicio + fim) // 2
        # Compara a string da lista em minúsculas
        nome_atual = lista_nomes[meio].lower()
        
        if nome_atual == alvo_normalizado:
            return meio
        elif nome_atual < alvo_normalizado:
            inicio = meio + 1
        else:
            fim = meio - 1
            
    return -1


nomes = ["Alberto", "Beatriz", "Carlos", "Daniela", "Eduardo"]
print(busca_binaria_nome(nomes, "dAnIeLa")) 
print(busca_binaria_nome(nomes, "Zeca"))    
```

```python
def busca_mais_proximo(lista, alvo):
    if not lista:
        return -1
        
    inicio = 0
    fim = len(lista) - 1
    
    # Busca Binária Padrão
    while inicio <= fim:
        meio = (inicio + fim) // 2
        
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
            
    
    # Se o alvo é maior que o último elemento
    if inicio >= len(lista):
        return len(lista) - 1
    # Se o alvo é menor que o primeiro elemento
    if fim < 0:
        return 0
        
    # Compara a diferença absoluta 
    dif_fim = abs(lista[fim] - alvo)
    dif_inicio = abs(lista[inicio] - alvo)
    
    return fim if dif_fim <= dif_inicio else inicio


nums = [10, 20, 30, 40, 50]

print(busca_mais_proximo(nums, 28)) 
print(busca_mais_proximo(nums, 12)) 
print(busca_mais_proximo(nums, 46)) 
```

```python
def contar_menores_que_x(lista, x):
    inicio = 0
    fim = len(lista) - 1
    
    while inicio <= fim:
        meio = (inicio + fim) // 2
        
        # Se o elemento no meio for menor que x, a resposta está à direita
        if lista[meio] < x:
            inicio = meio + 1
        # Se for maior ou igual, a resposta (ou o ponto de corte) está à esquerda
        else:
            fim = meio - 1
            
    # Ao final do loop, 'inicio' contém o número de elementos menores que x
    return inicio


numeros = [10, 20, 20, 30, 40, 50]

print(contar_menores_que_x(numeros, 25))
print(contar_menores_que_x(numeros, 10)) 
print(contar_menores_que_x(numeros, 60)) 
```

```python
def busca_matriz_ordenada(matriz, alvo):
    if not matriz or not matriz[0]:
        return (-1, -1)

    linhas = len(matriz)
    colunas = len(matriz[0])

    #  primeira linha, última coluna
    f = 0
    c = colunas - 1

    while f < linhas and c >= 0:
        valor_atual = matriz[f][c]
        
        if valor_atual == alvo:
            return (f, c)
        
        if valor_atual > alvo:
            c -= 1
     
        else:
            f += 1
            
    return (-1, -1)


matriz_4x4 = [
    [10, 20, 30, 40],
    [15, 25, 35, 45],
    [27, 29, 37, 48],
    [32, 33, 39, 50]
]

alvo = 37
print(f"Posição de {alvo}: {busca_matriz_ordenada(matriz_4x4, alvo)}") 
```

```python
def busca_array_rotacionado(nums, alvo):
    inicio = 0
    fim = len(nums) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if nums[meio] == alvo:
            return meio

        # metade e lado esquerdo
        if nums[inicio] <= nums[meio]:
         
            if nums[inicio] <= alvo < nums[meio]:
                fim = meio - 1  # esquerda
            else:
                inicio = meio + 1 # direita
        
        # 2. a metade direita 
        else:
            # O alvo está dentro desse intervalo ordenado?
            if nums[meio] < alvo <= nums[fim]:
                inicio = meio + 1 # Busca na direita
            else:
                fim = meio - 1    # Busca na esquerda

    return -1


lista_rotacionada = [4, 5, 6, 7, 0, 1, 2]
alvo = 0

resultado = busca_array_rotacionado(lista_rotacionada, alvo)
print(f"O índice do elemento {alvo} é: {resultado}") 
```

```python
import time
import random

def busca_binaria_rotacionada(nums, alvo):
    """
    Busca binária em lista rotacionada. Complexidade: O(log n).
    """
    if not nums:  # ista vazia
        return -1
    
    inicio, fim = 0, len(nums) - 1
    
    while inicio <= fim:
        meio = (inicio + fim) // 2
        
        if nums[meio] == alvo:
            return meio
        
        # Lado Esquerdo ordenado
        if nums[inicio] <= nums[meio]:
           
            if nums[inicio] <= alvo < nums[meio]:
                fim = meio - 1
            else:
                inicio = meio + 1
        # Lado Direito ordenado
        else:
            
            if nums[meio] < alvo <= nums[fim]:
                inicio = meio + 1
            else:
                fim = meio - 1
                
    return -1 #  alvo não encontrado

def busca_sequencial(nums, alvo):
    """
    Busca sequencial para fins de comparação. Complexidade: O(n).
    """
    for i in range(len(nums)):
        if nums[i] == alvo:
            return i
    return -1



def executar_teste():
    tamanho = 1_000_000  # 1 milhão de elementos para ver a diferença de tempo
    ponto_rotacao = random.randint(0, tamanho - 1)
    
   
    print(f"Gerando lista de {tamanho} elementos e rotacionando no índice {ponto_rotacao}...")
    lista_base = sorted([random.randint(0, tamanho) for _ in range(tamanho)])
    lista_rotacionada = lista_base[ponto_rotacao:] + lista_base[:ponto_rotacao]
    
    alvo = random.choice(lista_rotacionada)
    
    print("-" * 50)
    
    # Teste 1 Busca Sequencial
    t_inicio = time.time()
    res_seq = busca_sequencial(lista_rotacionada, alvo)
    t_fim = time.time()
    tempo_seq = t_fim - t_inicio
    print(f"Busca Sequencial | Índice: {res_seq} | Tempo: {tempo_seq:.6f}s")
    
    # Teste 2 Busca Binária Modificada
    t_inicio = time.time()
    res_bin = busca_binaria_rotacionada(lista_rotacionada, alvo)
    t_fim = time.time()
    tempo_bin = t_fim - t_inicio
    print(f"Busca Binária    | Índice: {res_bin} | Tempo: {tempo_bin:.6f}s")
    
    # Comparação
    if tempo_bin > 0:
        print(f"\nA Busca Binária foi {tempo_seq / tempo_bin:.2f}x mais rápida!")
    
    # Teste 3: Alvo não encontrado
    print(f"\nTeste Alvo Inexistente (-999): {busca_binaria_rotacionada(lista_rotacionada, -999)}")
    # Teste 4: Lista Vazia
    print(f"Teste Lista Vazia []: {busca_binaria_rotacionada([], 10)}")

if __name__ == "__main__":
    executar_teste()
```
