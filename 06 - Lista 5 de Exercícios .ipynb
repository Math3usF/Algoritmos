```python
# Tratamento de exceções básicas: Escreva um programa que peça ao usuário dois números e faça a divisão do primeiro pelo segundo. 
# Se o usuário inserir um valor inválido ou tentar dividir por zero, o programa deve exibir uma mensagem de erro apropriada.

n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))

try:
    resultado = n1/n2
    print(f'O resultado da divisão é: {resultado}')
except ZeroDivisionError: #erro de divisão por zero
    print("Error, não é possível dividir por zero.")
```

```python
# Capturando exceções múltiplas: Crie um programa que peça ao usuário o nome de uma cor e mostre seu valor em RGB de acordo com um dicionário
# pré-definido. O programa deve tratar exceções caso o nome da cor não exista no dicionário.

cores = {"vermelho":(255, 0, 0), "verde":(0, 255, 0), "azul":(0, 0, 255)}
cor = input("Digite a cor desejada: ")

try: 
    if cor in cores:
        print(f" A cor {cor} tem seu valor em RGB: {cores[cor]}")
    if cor not in cores:
        print("Error, cor não encontrada.")
except KeyError: #erro por chave não encontrada
    print("Error, cor não encontrada.")
```

```python
# Bloco else e finally: Escreva um programa que solicite um número ao usuário. Se o número for maior que 10, exiba uma mensagem dizendo que o
# número é válido. Utilize o bloco else para imprimir que o programa foi executado com sucesso, e o bloco finally para imprimir "Programa
# encerrado".

try:
    n = int(float(input("Digite um número: ")))

    if n > 10:
        print("O número é válido.")
    else: 
        print("O número é inválido.")
except ValueError: #erro por valor inválido
    print("Error!! Por favor, digite um número.")        
finally:
    print("Programa finalizado.")

```

```python
# Exceções personalizadas: Escreva uma função que verifica se uma senha # possui no mínimo 8 caracteres e pelo menos um número. Se a senha não
# atender aos requisitos, levante uma exceção com uma mensagem # personalizada. Trate a exceção e mostre a mensagem ao usuário.

def senha_valida(senha):
    if len(senha) < 8:
        raise ValueError("A senha deve conter pelo menos 8 caracteres.") #raise é usado para lançar um erro de exceção personalizada. Pode utilizar também um print para mostrar a mensagem de erro.
    if not any(char.isdigit() for char in senha): #any (char.isdigit() for char in senha) verifica se tem pelo menos um dígito na senha.
        raise ValueError("A senha deve conter pelo menos um número.")
    if not any(char.isalpha() for char in senha): #any (char.isalpha() for char in senha) verifica se tem pelo menos uma letra na senha.
        raise ValueError("A senha deve conter pelo menos uma letra.")
    return True #atenção com Return, se quiser que tenha um retorno ele precisa ser colocado no final da função como True, senão a função não retorna nada.

senha = input("Digite uma senha: ")        
try:
    if senha_valida(senha):
        print("Senha válida.")
except ValueError as e:
    print(f"Error!! {e}")
```

```python
saldo = float(input("Digite o saldo da conta: R$ "))
saque = float(input("Digite o valor do saque: R$ "))

def sacar(saldo, saque): 
    if saldo < 0: 
        print("Error!! O saldo não pode ser negativo.")
    if saque > saldo:
        print("Saldo insuficiente para realizar o saque.")
    return saldo - saque
try:   
    novo_saldo = sacar(saldo, saque)
    print(f"Saque realizado com sucesso. Novo saldo: R$ {novo_saldo}")
except ValueError as e:
    print(f"Error!! {e}")
```
